```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import time

def automacao_aulas():
    print("🤖 Iniciando o robô...")
    
    # Abre o navegador Chrome (você precisa ter o Google Chrome instalado)
    driver = webdriver.Chrome()
    
    # Substitua pelo link da página de login ou do curso
    url_do_curso = "https://portalhashtag.com/cursos/1772107635073x822895358782933100?modulo=1732572586255x381537996406748000" 
    driver.get(url_do_curso)
    
    print("\n" + "="*50)
    print("ATENÇÃO: Uma janela do Chrome foi aberta.")
    print("1. Faça o seu login.")
    print("2. Navegue até a página exata onde estão as aulas do módulo.")
    print("3. Volte aqui neste terminal e aperte ENTER para começar os cliques.")
    print("="*50 + "\n")
    
    # Pausa o script até você dar o ok
    input("Aperte ENTER quando estiver na página certa...")
    
    print("🔍 Buscando os botões de aula...")
    
    # Busca os elementos usando o mesmo seletor que descobrimos no Bubble
    seletor = ".clickable-element.bubble-element.Group"
    botoes = driver.find_elements(By.CSS_SELECTOR, seletor)
    
    if len(botoes) == 0:
        print("⚠️ Nenhum botão encontrado. O site pode ter carregado diferente ou o seletor está errado.")
        driver.quit()
        return

    print(f"🚀 Encontrados {len(botoes)} elementos clicáveis. Iniciando...")

    for i, botao in enumerate(botoes):
        try:
            print(f"⏳ Marcando item {i + 1} de {len(botoes)}...")
            
            # Usando JavaScript interno do Selenium para clicar (evita erros se o botão estiver fora da tela)
            driver.execute_script("arguments[0].click();", botao)
            
            # Pausa de 2.5 segundos para a plataforma registrar a conclusão
            time.sleep(2.5)
            
        except Exception as e:
            print(f"❌ Erro ao clicar no item {i + 1}: {e}")

    print("✅ Processo finalizado!")
    
    # Opcional: fechar o navegador no final
    # driver.quit() 

if __name__ == "__main__":
    automacao_aulas()
```
